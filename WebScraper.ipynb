```python
import os
import asyncio # Import asyncio for running async functions
from playwright.async_api import async_playwright # Changed to async_playwright
from bs4 import BeautifulSoup

# Install system dependencies for Playwright
!apt-get update
!apt-get install -y libatk-bridge2.0-0 libgtk-3-0 libgdk-pixbuf2.0-0 libjpeg-turbo8 libxss1 libgbm-dev

async def scrape_ucd_page(url, output_filename): # Changed to async def
    print(f"Scraping dynamic page: {url}")
    async with async_playwright() as p: # Changed to async with async_playwright
        # Launch a headless browser
        browser = await p.chromium.launch(headless=True) # Added await
        page = await browser.new_page() # Added await

        # Navigate to the UCD dining link and wait until the network is quiet
        await page.goto(url, wait_until="networkidle") # Added await

        # Get the fully rendered HTML content
        html_content = await page.content() # Added await
        await browser.close() # Added await

    # Use BeautifulSoup to clean the data
    soup = BeautifulSoup(html_content, "html.parser")

    # PROJECT REQUIREMENT: Remove navigation text, headers, and ads
    for element in soup(["nav", "header", "footer", "script", "style", "aside"]):
        element.decompose()

    # Target the main content container where UCD stores article/menu text
    main_content = soup.find("main") or soup.find("div", {"id": "content"}) or soup.body

    # Clean up whitespace
    text = main_content.get_text(separator="\n")
    cleaned_lines = [line.strip() for line in text.splitlines() if line.strip()]
    final_text = "\n".join(cleaned_lines)

    # Save the structured text asset
    os.makedirs("data/raw", exist_ok=True)
    with open(f"data/raw/{output_filename}.txt", "w", encoding="utf-8") as f:
        f.write(f"Source URL: {url}\n\n" + final_text)
    print(f"Successfully saved to data/raw/{output_filename}.txt")

# Example execution for your source #1 and #3
# To run async functions at the top level in Colab, we can define an async main function.
async def run_scrapers():
    await scrape_ucd_page("https://housing.ucdavis.edu/dining/dining-commons/segundo/", "ucd_segundo_dc")
    await scrape_ucd_page("https://housing.ucdavis.edu/dining/the-gunrock/", "gunrock")
    await scrape_ucd_page("https://housing.ucdavis.edu/dining/food-trucks/", "food_trucks")
    await scrape_ucd_page("https://housing.ucdavis.edu/dining/sage-street/", "sage_street")
    await scrape_ucd_page("https://housing.ucdavis.edu/dining/latitude-market/", "latitude_market")
    await scrape_ucd_page("https://housing.ucdavis.edu/dining/latitude/", "latitude_restaurant")

# Run the async main function
await run_scrapers()
```
